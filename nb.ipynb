```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import squarify
from matplotlib.patches import Patch
```

```python
# Requisição dos csv's e criandos os df's para a plotagem dos gráficos
senna_df_qualy = pd.read_csv('data-csv/qualifying-results/qualifying-results_ayrton-senna.csv')
prost_df_qualy = pd.read_csv('data-csv/qualifying-results/qualifying-results_alain-prost.csv')
mansell_df_qualy = pd.read_csv('data-csv/qualifying-results/qualifying-results_nigel-mansell.csv')
piquet_df_qualy = pd.read_csv('data-csv/qualifying-results/qualifying-results_nelson-piquet.csv')
berger_df_qualy = pd.read_csv('data-csv/qualifying-results/qualifying-results_gerhard-berger.csv')
lauda_df_qualy = pd.read_csv('data-csv/qualifying-results/qualifying-results_niki-lauda.csv')
schumacher_df_qualy = pd.read_csv('data-csv/qualifying-results/qualifying-results_michael-schumacher.csv')
hamilton_df_qualy = pd.read_csv('data-csv/qualifying-results/qualifying-results_lewis-hamilton.csv')
vetell_df_qualy = pd.read_csv('data-csv/qualifying-results/qualifying-results_sebastian-vettel.csv')
alonso_df_qualy = pd.read_csv('data-csv/qualifying-results/qualifying-results_fernando-alonso.csv')
verstappen_df_qualy = pd.read_csv('data-csv/qualifying-results/qualifying-results_max-verstappen.csv')
cecotto_df_qualy = pd.read_csv('data-csv/qualifying-results/qualifying-results_johnny-cecotto.csv')
johansson_df_qualy = pd.read_csv('data-csv/qualifying-results/qualifying-results_stefan-johansson.csv')
dumfries_df_qualy = pd.read_csv('data-csv/qualifying-results/qualifying-results_johnny-dumfries.csv')
angelis_df_qualy = pd.read_csv('data-csv/qualifying-results/qualifying-results_elio-de-angelis.csv')
nakajima_df_qualy = pd.read_csv('data-csv/qualifying-results/qualifying-results_satoru-nakajima.csv')
```

### Senna em Classificatórias

```python
# Arrays que contém anos que serão interessantes para a análise dos dados
years_1988_1994 = [1988, 1989, 1990, 1991, 1992, 1993, 1994]
years_until_1991 = [1988, 1989, 1990, 1991]
years_wo_1994 = [1988, 1989, 1990, 1991, 1992, 1993]

# Aplicando os anos nos df's
senna_data_1988_1994 = senna_df_qualy[senna_df_qualy['year'].isin(years_1988_1994)]
senna_data_1988_1994 = senna_data_1988_1994.reset_index()
prost_data_1988_1991 = prost_df_qualy[prost_df_qualy['year'].isin(years_until_1991)]
prost_data_1993 = prost_df_qualy[prost_df_qualy['year'] == 1993]
mansell_data_1988_1992 = mansell_df_qualy[mansell_df_qualy['year'].isin(years_wo_1994)]
piquet_data_1988_1991 = piquet_df_qualy[piquet_df_qualy['year'].isin(years_until_1991)]
berger_data_1988_1993 = berger_df_qualy[berger_df_qualy['year'].isin(years_wo_1994)]
berger_data_1994 = berger_df_qualy[berger_df_qualy['year'] == 1994].head(3)
berger_data_1988_1994 = pd.concat([berger_data_1988_1993, berger_data_1994], axis=0)

# Puramente para evitar o erro de SettingWithCopyWarning
senna_data_1988_1994 = senna_data_1988_1994.copy()
prost_data_1988_1991 = prost_data_1988_1991.copy()
prost_data_1993 = prost_data_1993.copy()
mansell_data_1988_1992 = mansell_data_1988_1992.copy()
piquet_data_1988_1991 = piquet_data_1988_1991.copy()
berger_data_1988_1994 = berger_data_1988_1994.copy()

# Criando a coluna 'race' ('year') para a plotagem futura
senna_data_1988_1994['year_race'] = senna_data_1988_1994['race'] + " (" + senna_data_1988_1994['year'].astype(str) + ')'
prost_data_1988_1991['year_race'] = prost_data_1988_1991['race'] + " (" + prost_data_1988_1991['year'].astype(str) + ')'
prost_data_1993['year_race'] = prost_data_1993['race'] + " (" + prost_data_1993['year'].astype(str) + ')'
mansell_data_1988_1992['year_race'] = mansell_data_1988_1992['race'] + " (" + mansell_data_1988_1992['year'].astype(str) + ')'
piquet_data_1988_1991['year_race'] = piquet_data_1988_1991['race'] + " (" + piquet_data_1988_1991['year'].astype(str) + ')'
berger_data_1988_1994['year_race'] = berger_data_1988_1994['race'] + " (" + berger_data_1988_1994['year'].astype(str) + ')'
```

```python
fig, ax = plt.subplots(figsize=(28,8))

# Plotagem da classificação de todos os pilotos que são interessantes para análise
plt.plot(senna_data_1988_1994['year_race'], senna_data_1988_1994['position'], marker='o', linestyle='-', color='#e91c0d', markersize=3, label='Ayrton Senna')
plt.plot(prost_data_1988_1991['year_race'], prost_data_1988_1991['position'], marker='o', linestyle='-', color='#0066CC', markersize=3, label='Alain Prost')
plt.plot(prost_data_1993['year_race'], prost_data_1993['position'], marker='o', linestyle='-', color='#0066CC', markersize=2, label='')
plt.plot(mansell_data_1988_1992['year_race'], mansell_data_1988_1992['position'], marker='o', linestyle='-', color='#ffb700', markersize=3, label='Nigel Mansell')
plt.plot(piquet_data_1988_1991['year_race'], piquet_data_1988_1991['position'], marker='o', linestyle='-', color='#38812F', markersize=3, label='Nelson Piquet')
plt.plot(berger_data_1988_1994['year_race'], berger_data_1988_1994['position'], marker='o', linestyle='-', color='#5a1b1b', markersize=3, label='Gerhard Beger')

# Configurações do gráfico como título e estética
plt.title('Ayrton Senna vs. Outros Pilotos - Resultados Qualificatórios (1988-1994)', fontsize=14)
plt.xlabel('Corrida (Ano)', fontsize=12)
plt.ylabel('Posição de Qualificação', fontsize=12)
plt.xticks(rotation=90, fontsize=10)
plt.yticks([1, 5, 10, 15, 20, 25], fontsize=10)
plt.gca().invert_yaxis()
plt.grid(axis='y')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Coloração das temporadas em que Senna se consagrou campeão da F1
plt.axvspan(-0.5, len(senna_data_1988_1994[senna_data_1988_1994['year'] == 1988])-0.5, facecolor='#f7f1d0', alpha=0.5)

start_index_wc = senna_data_1988_1994[senna_data_1988_1994['year'] >= 1991].index[0] - senna_data_1988_1994[senna_data_1988_1994['year'] == 1989].index[0]
end_index_wc = start_index_wc + len(senna_data_1988_1994[senna_data_1988_1994['year'] == 1991])
plt.axvspan(start_index_wc, end_index_wc-0.5, facecolor='#f7f1d0', alpha=0.5)

start_index_wc = senna_data_1988_1994[senna_data_1988_1994['year'] >= 1992].index[0] - senna_data_1988_1994[senna_data_1988_1994['year'] == 1989].index[0]
end_index_wc = start_index_wc + len(senna_data_1988_1994[senna_data_1988_1994['year'] == 1992])
plt.axvspan(start_index_wc, end_index_wc-0.5, facecolor='#f7f1d0', alpha=0.5)

# Coloração da temporada em que houve o domínio da Williams Racing
start_index_williams = senna_data_1988_1994[senna_data_1988_1994['year'] >= 1994].index[0] - senna_data_1988_1994[senna_data_1988_1994['year'] == 1990].index[0]
end_index_williams = len(senna_data_1988_1994['year_race'])-1
plt.axvspan(start_index_williams-0.5, end_index_williams+0.5, facecolor='#b4d6f8', alpha=0.5)

# Configurações da legenda
legend_elements = [
    Patch(facecolor='#8BC1F7', edgecolor='#b4d6f8', label='Domínio da Williams Racing'),
    Patch(facecolor='#f5e9a3', edgecolor='#f7f1d0', label='Campeonatos Mundiais do Senna'),
]
legend = plt.legend(handles=plt.gca().get_legend_handles_labels()[0] + legend_elements, loc='lower left')

# Plotagem
plt.tight_layout()
plt.show()
```

### Toleman

```python
# Separando os pilotos e o ano da Toleman
senna_data_1984 = senna_df_qualy[senna_df_qualy['year'] == 1984]
cecotto_data_1984 = cecotto_df_qualy[cecotto_df_qualy['year'] == 1984]
johansson_data_1984 = johansson_df_qualy[johansson_df_qualy['year'] == 1984]

# Puramente para evitar o erro SettingWithCopyWarning
senna_data_1984 = senna_data_1984.copy()
cecotto_data_1984 = cecotto_data_1984.copy()
johansson_data_1984 = johansson_data_1984.copy()

# Criando a coluna 'race' ('year') para a plotagem futura
senna_data_1984['year_race'] = senna_data_1984['race'] + " (" + senna_data_1984['year'].astype(str) + ')'
cecotto_data_1984['year_race'] = cecotto_data_1984['race'] + " (" + cecotto_data_1984['year'].astype(str) + ')'
johansson_data_1984['year_race'] = johansson_data_1984['race'] + " (" + johansson_data_1984['year'].astype(str) + ')'

# Registro das vezes em que Senna ficou em uma posição pior que seu companheiro
loses_toleman = []

for index, (senna_row, cecotto_row) in enumerate(zip(senna_data_1984.iterrows(), cecotto_data_1984.iterrows())):
    senna_position = senna_row[1]['position']
    cecotto_position = cecotto_row[1]['position']

    if senna_position > cecotto_position:
        loses_toleman.append(index)
```

```python
plt.figure(figsize=(10,5))

# Plotagem das classificações dos pilotos
plt.plot(senna_data_1984['year_race'], senna_data_1984['position'], marker='o', linestyle='-', color='#e91c0d', markersize=3, label='Ayrton Senna')
plt.plot(cecotto_data_1984['year_race'], cecotto_data_1984['position'], marker='o', linestyle='-', color='#0066CC', markersize=3, label='Johnny Cecotto')
plt.plot(johansson_data_1984['year_race'], johansson_data_1984['position'], marker='o', linestyle='-', color='#ffb700', markersize=3, label='Stefan Johansson')

# Configurações do gráfico como título e estética
plt.title('Ayrton Senna vs. Companheiros de Equipe da Toleman - Resultados Qualificatórios', fontsize=14)
plt.xlabel('Corrida (Ano)', fontsize=12)
plt.ylabel('Posição de Qualificação', fontsize=12)
plt.xticks(rotation=90, fontsize=10)
plt.yticks([1, 5, 10, 15, 20, 25], fontsize=10)
plt.gca().invert_yaxis()
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Marcação em vermelho das corridas em que Senna ficou em uma posição pior que seu companheiro
for lose in loses_toleman:
    plt.axvspan(lose-0.5, lose+0.5, facecolor='#f8b4b4', alpha=0.5)

# Configurações de legenda
legend_elements = [
    Patch(facecolor='#f8b4b4', edgecolor='#f8b4b4', label="Derrotas do Senna"),
]
legend = plt.legend(handles=plt.gca().get_legend_handles_labels()[0] + legend_elements, loc='upper left')

# Plotagem
plt.tight_layout()
plt.show()
```

### McLaren

```python
# Separação dos anos em que Senna correu pela McLaren
mclaren_years = [1988, 1989, 1990, 1991, 1992, 1993]
senna_data_1988_1993 = senna_df_qualy[senna_df_qualy['year'].isin(mclaren_years)]
senna_data_1992_1993 = senna_df_qualy[senna_df_qualy['year'].isin([1992, 1993])]
berger_data_1992_1993 = berger_df_qualy[berger_df_qualy['year'].isin([1992, 1993])]

# Puramente para evitar o erro SettingWithCopyWarning
senna_data_1988_1993 = senna_data_1988_1993.copy()
berger_data_1992_1993 = berger_data_1992_1993.copy()

# Criando a coluna 'race' ('year') para a plotagem futura
senna_data_1988_1993['year_race'] = senna_data_1988_1993['race'] + " (" + senna_data_1988_1993['year'].astype(str) + ')'
berger_data_1992_1993['year_race'] = berger_data_1992_1993['race'] + " (" + berger_data_1992_1993['year'].astype(str) + ')'

# Registro das vezes em que Senna ficou em uma posição pior que seu companheiro
loses_mclaren = []

for index, (senna_row, prost_row) in enumerate(zip(senna_data_1988_1993.iterrows(), prost_data_1988_1991.iterrows())):
    senna_position = senna_row[1]['position']
    prost_position = prost_row[1]['position']

    if senna_position > prost_position:
        loses_mclaren.append(index)

for index, (senna_row, berger_row) in enumerate(zip(senna_data_1992_1993.iterrows(), berger_data_1992_1993.iterrows())):
    senna_position = senna_row[1]['position']
    berger_position = berger_row[1]['position']

    if senna_position > berger_position:
        lose_index = index + len(prost_data_1988_1991)
        loses_mclaren.append(lose_index+1)
```

```python
plt.figure(figsize=(18,5))

# Plotagem das classificações dos pilotos
plt.plot(senna_data_1988_1993['year_race'], senna_data_1988_1993['position'], marker='o', linestyle='-', color='#e91c0d', markersize=2, label='Ayrton Senna')
plt.plot(prost_data_1988_1991['year_race'], prost_data_1988_1991['position'], marker='o', linestyle='-', color='#0066CC', markersize=2, label='Alain Prost')
plt.plot(berger_data_1992_1993['year_race'], berger_data_1992_1993['position'], marker='o', linestyle='-', color='#ffb700', markersize=2, label='Gerhard Berger')

# Configurações de gráfico como título e estética
plt.title('Ayrton Senna vs. Companheiros de Equipe da McLaren - Resultados de Qualificação', fontsize=14)
plt.xlabel('Corrida (Ano)', fontsize=12)
plt.ylabel('Posição de Qualificação', fontsize=12)
plt.xticks(rotation=90, fontsize=10)
plt.yticks([1, 5, 10, 15, 20, 25], fontsize=10)
plt.gca().invert_yaxis()
plt.grid(axis='y')
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Marcação em vermelho das corridas em que Senna ficou em uma posição pior que seu companheiro
for lose in loses_mclaren:
    plt.axvspan(lose-0.5, lose+0.5, facecolor='#f8b4b4', alpha=0.5)

# Configurações de legenda
legend_elements = [
    Patch(facecolor='#f8b4b4', edgecolor='#f8b4b4', label="Derrotas do Senna"),
]
legend = plt.legend(handles=plt.gca().get_legend_handles_labels()[0] + legend_elements, loc='lower left')

# Plotagem
plt.tight_layout()
plt.show()
```

### Rain Master

```python
# Requisição dos csv's e criandos os df's para a plotagem dos gráficos
df_monaco_1984_gap = pd.read_csv('./data-csv/rain-races/1984-monaco-gap.csv')
df_monaco_1984_laptime = pd.read_csv('./data-csv/rain-races/1984-monaco-laptime.csv')

# Função para converter tempo em segundos
def to_seconds(val):
    if pd.isna(val):
        return None
    try:
        if isinstance(val, str) and ':' in val:
            m, s = val.split(':')
            return int(m) * 60 + float(s)
        return float(val)
    except:
        return None

# Tratamento de dados para as voltas e distância
laps = df_monaco_1984_laptime.columns[1:]
gaps = [col for col in df_monaco_1984_gap.columns if col.startswith('gap')]

# Converter os tempos
df_monaco_1984_laptime[laps] = df_monaco_1984_laptime[laps].map(to_seconds)

# Filtrar Senna e Prost
senna = df_monaco_1984_laptime[df_monaco_1984_laptime['driver'].str.contains('SEN', case=False)].iloc[0]
prost = df_monaco_1984_laptime[df_monaco_1984_laptime['driver'].str.contains('PRO', case=False)].iloc[0]

# Dados prontos para plotagem
senna_times = senna[laps].values
prost_times = prost[laps].values
lap_numbers = list(range(1, len(laps) + 1))

# Extrair linhas de Senna e Prost
senna_row = df_monaco_1984_gap[df_monaco_1984_gap['driver'] == 'SEN'][gaps].iloc[0]
prost_row = df_monaco_1984_gap[df_monaco_1984_gap['driver'] == 'PRO'][gaps].iloc[0]

# Converter gaps
senna_gap = senna_row.map(to_seconds).tolist()
prost_gap = prost_row.map(to_seconds).tolist()
prost_base = max([v for v in prost_gap if v is not None])

# Voltas (x-axis)
laps_gap = list(range(1, len(senna_gap) + 1))

# Ajustar Senna: gap real somado ao valor base
senna_adjusted = [prost_base + val if val is not None else None for val in senna_gap]
```

```python
fig, ax = plt.subplots(figsize=(18, 6))

# Plotagem dos tempos dos pilotos
plt.plot(lap_numbers, senna_times, label='Ayrton Senna', color='#e91c0d', linewidth=2)
plt.plot(lap_numbers, prost_times, label='Alain Prost', color='#0066CC', linewidth=2)

# COnfigurações do gŕafico como título e estética
plt.title('Ayrton Senna vs. Alain Prost - GP de Mônaco 1984 - Tempos de Volta', fontsize=16)
plt.xlabel('Volta', fontsize=12)
plt.ylabel('Tempo de Volta (s)', fontsize=12)
plt.xticks(ticks=range(1, 32), fontsize=10)
plt.yticks(fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Configuração de legenda
legend_elements = [
    Patch(facecolor='#e91c0d', label='Ayrton Senna'),
    Patch(facecolor='#0066CC', label='Alain Prost'),
]
plt.legend(handles=legend_elements, loc='upper right', frameon=False, fontsize=11)

# Plotagem
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(18, 6))

# Plotagem da distância de Senna para Prost em segundos
ax.plot(laps_gap, senna_adjusted, label='Ayrton Senna', color='#e91c0d', linewidth=2)

# Adicionar valores da distância acima da linha do Senna a cada 5 voltas e nas 6 últimas voltas
for i, (lap, gap) in enumerate(zip(laps_gap, senna_adjusted)):
    if ((lap - 1) % 5 == 0 or lap > len(laps) - 6) and gap is not None:
        ax.text(
            lap, 
            gap - 2.5,
            f"{senna_gap[i]:.1f}" if senna_gap[i] is not None else "", 
            color='black', 
            fontsize=10, 
            ha='center'
        )

# Configuração do gráfico como título e estética
ax.set_title('Diferença de Tempo de Senna para Prost - GP de Mônaco 1984', fontsize=16)
ax.set_xlabel('Volta', fontsize=12)
ax.set_ylabel('Distância em relação ao Prost (s)', fontsize=12)
max_y = max([v for v in senna_adjusted if v is not None])
ax.set_yticks(np.arange(0, max_y + 5, 5))
ax.tick_params(axis='x', labelsize=10)
ax.tick_params(axis='y', labelsize=10)
ax.grid(axis='y', linestyle='--', alpha=0.3)
ax.legend(frameon=False, fontsize=11)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Plotagem
plt.tight_layout()
plt.show()
```
